```python
import math

def minimax(depth, node, maximizing_player, values):
    # Base case: leaf node
    if depth == 3:
        return values[node]

    if maximizing_player:
        best = -math.inf

        for i in range(2):
            val = minimax(depth + 1, node * 2 + i,
                          False, values)
            best = max(best, val)

        return best

    else:
        best = math.inf

        for i in range(2):
            val = minimax(depth + 1, node * 2 + i,
                          True, values)
            best = min(best, val)

        return best


# Leaf node values
values = [3, 5, 6, 9, 1, 2, 0, -1]

result = minimax(0, 0, True, values)

print("Optimal value:", result)
```
